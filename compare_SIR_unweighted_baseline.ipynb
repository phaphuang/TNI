```python
from baseline.compute_baseline import baseline
import numpy as np
import networkx as nx

from utils import cal_Kendall_tau_coefficient, load_graph_data
from baseline.tpks import compute_TPKS
from baseline.hvgc import compute_hvgc
from baseline.glc import compute_glc
from tni import compute_TNI

import os
```

```python
# Load networks
selected_unweighted_networks = [
    "01_Karate", "02_Dolphins", "03_Polbooks", 
    "04_Jazz", "05_WM2", "06_Power",
    "07_EEC", "08_Email", "09_Blogs",
    "10_Router", "11_PG", "12_WikiVote",
]

# Set truncation radius (R=2)
R_list = [1, 2]

# Define 
nums = 11
```

```python
root_folder = "data"
network_type = "unweighted"
folder_name = root_folder + "/" + network_type
```

```python
for R in R_list:
    # Scores comparison
    for network_name in selected_unweighted_networks:
        print(f"\nR={R}\n{network_name} Network:")
        
        # Calculate scores
        beta_c, dc_scores, dc_and_scores, dcgm_scores, \
        ggc_scores, dcgm_plus_scores, dcgm_plus_plus_scores = baseline(folder_name, network_name, R)

        #print(f"\nDC_AND Scores: {dc_and_scores} \nDCGM Scores: {dcgm_scores} \nGCC Scores: {gcc_scores} \nDCGM+ Scores: {dcgm_plus_scores} \nDCGM++ Scores: {dcgm_plus_plus_scores}")
        
        if 'karate' in network_name.lower():
            G = nx.karate_club_graph()
        else:
            G = load_graph_data(folder_name=folder_name, name=network_name)

        tpks_scores = compute_TPKS(G)
        hvgc_scores = compute_hvgc(G, R=R)
        glc_scores = compute_glc(G)
        cm_scores, tni_scores  = compute_TNI(G, R=R, network_type=network_type)

        # Kendall's Tau comparison
        X1 = list(dc_scores.values())
        X2 = list(dc_and_scores.values())
        X3 = list(cm_scores.values())
        X4 = list(tpks_scores.values())
        X5 = list(dcgm_scores.values())
        X6 = list(ggc_scores.values())
        X7 = list(dcgm_plus_scores.values())
        X8 = list(dcgm_plus_plus_scores.values())
        X9 = list(hvgc_scores.values())
        X10 = list(glc_scores.values())
        X11 = list(tni_scores.values())

        # calculate with different beta
        beta_list = np.linspace(0.5, 1.5, nums) * beta_c
        tau_list1 = np.zeros(nums)
        tau_list2 = np.zeros(nums)
        tau_list3 = np.zeros(nums)
        tau_list4 = np.zeros(nums)
        tau_list5 = np.zeros(nums)
        tau_list6 = np.zeros(nums)
        tau_list7 = np.zeros(nums)
        tau_list8 = np.zeros(nums)
        tau_list9 = np.zeros(nums)
        tau_list10 = np.zeros(nums)
        tau_list11 = np.zeros(nums)

        # SIR ranking
        SR = np.loadtxt(f"./sir_ranking_{network_type}_results/sir_ranking_" + network_name + ".csv", delimiter=',', dtype=np.float64)
        
        out_folder = f"kendall_tau_{network_type}_r{str(R).lower()}_results"
        os.makedirs(out_folder, exist_ok=True)

        outf = open(f"./{out_folder}/kendall_tau_" + network_name + ".dat", "w")
        for i, beta in enumerate(beta_list):
            SRi = SR[:, i+1]
            Y = list(SRi)
            tau_list1[i] = cal_Kendall_tau_coefficient(X1, Y)
            tau_list2[i] = cal_Kendall_tau_coefficient(X2, Y)
            tau_list3[i] = cal_Kendall_tau_coefficient(X3, Y)
            tau_list4[i] = cal_Kendall_tau_coefficient(X4, Y)
            tau_list5[i] = cal_Kendall_tau_coefficient(X5, Y)
            tau_list6[i] = cal_Kendall_tau_coefficient(X6, Y)
            tau_list7[i] = cal_Kendall_tau_coefficient(X7, Y)
            tau_list8[i] = cal_Kendall_tau_coefficient(X8, Y)
            tau_list9[i] = cal_Kendall_tau_coefficient(X9, Y)
            tau_list10[i] = cal_Kendall_tau_coefficient(X10, Y)
            tau_list11[i] = cal_Kendall_tau_coefficient(X11, Y)

            outf.write(str(beta/beta_c) + " " + str(tau_list1[i]) + " " + str(tau_list2[i]) +
                        " " + str(tau_list3[i]) + " " + str(tau_list4[i]) +
                        " " + str(tau_list5[i]) + " " + str(tau_list6[i]) + 
                        " " + str(tau_list7[i]) + " " + str(tau_list8[i]) + 
                        " " + str(tau_list9[i]) + " " + str(tau_list10[i]) +
                        " " + str(tau_list11[i]) + "\n")

        outf.close()
```
